```python
import os, requests, json as _json

DATA_DIR = "/content/zoolake_data"
os.makedirs(DATA_DIR, exist_ok=True)

# CKAN portals like opendata.eawag.ch expose a stable JSON API even when the raw
# download links (and resource UUIDs) shift over time or block plain wget/robots.
# Resolve the real, current download URLs from the API instead of hardcoding them.
CKAN_BASE = "https://opendata.eawag.ch"
DATASET_ID = "1aee0a2a-e1a7-4910-8151-4fc67c15dc63"  # ZooLake dataset package id

api_url = f"{CKAN_BASE}/api/3/action/package_show?id={DATASET_ID}"
headers = {"User-Agent": "Mozilla/5.0 (research download script)"}

resp = requests.get(api_url, headers=headers, timeout=30)
resp.raise_for_status()
pkg = resp.json()["result"]

resources = {r["name"]: r["url"] for r in pkg["resources"]}
print("Resources found on the dataset page:")
for name, url in resources.items():
    print(" -", name, "->", url)

def find_resource_url(keyword):
    for name, url in resources.items():
        if keyword.lower() in name.lower():
            return url
    raise KeyError(f"No resource matching '{keyword}' found — inspect the printed list above and set the URL manually.")

ZOOLAKE_URL = find_resource_url("ZooLake2")
OOD_URL = find_resource_url("OOD_data")

def download(url, out_path):
    r = requests.get(url, headers=headers, stream=True, timeout=120)
    r.raise_for_status()
    with open(out_path, "wb") as f:
        for chunk in r.iter_content(chunk_size=1 << 20):
            f.write(chunk)
    print(f"downloaded {out_path} ({os.path.getsize(out_path) / 1e6:.1f} MB)")

download(ZOOLAKE_URL, f"{DATA_DIR}/ZooLake2.zip")
download(OOD_URL, f"{DATA_DIR}/OOD_data.zip")

!cd {DATA_DIR} && unzip -q -o ZooLake2.zip -d zoolake2 && echo "unzipped ZooLake2"
!cd {DATA_DIR} && unzip -q -o OOD_data.zip -d ood && echo "unzipped OOD"
```

```python
import pandas as pd
import numpy as np
import glob

ROOT = f"{DATA_DIR}/zoolake2/ZooLake2.0"
split_pkl = glob.glob(f"{DATA_DIR}/zoolake2/**/Files_used_for_training_testing.pickle", recursive=True)[0]

# Read the pickle file, which returns a list of Series, and convert it to a DataFrame.
loaded_data_list = pd.read_pickle(split_pkl)
split_df = pd.DataFrame(pd.concat(loaded_data_list))
print(split_df.head())
print(split_df.columns.tolist())
```

```python
class_dirs = sorted([d for d in glob.glob(f"{ROOT}/*") if os.path.isdir(d)])
print(f"{len(class_dirs)} classes found")
print([os.path.basename(d) for d in class_dirs][:10], "...")

# Classes to treat as an explicit REJECT bucket rather than ordinary taxa
REJECT_CLASSES = {"dirt", "unknown", "unknown_plankton", "unknown plankton"}
```

```python
import torch
from torch.utils.data import Dataset
from PIL import Image
import torchvision.transforms as T

class ZooLakeFusionDataset(Dataset):
    def __init__(self, df, class_to_idx, feature_cols, transform=None, feature_scaler=None):
        self.df = df.reset_index(drop=True)
        self.class_to_idx = class_to_idx
        self.feature_cols = feature_cols
        self.transform = transform
        self.feature_scaler = feature_scaler  # (mean, std) tuple, fit on TRAIN split only

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        img = Image.open(row["filepath"]).convert("RGB")
        if self.transform:
            img = self.transform(img)

        feats = row[self.feature_cols].values.astype(np.float32)
        feats = np.nan_to_num(feats, nan=0.0, posinf=0.0, neginf=0.0)
        if self.feature_scaler is not None:
            mean, std = self.feature_scaler
            feats = (feats - mean) / (std + 1e-6)
        feats = torch.tensor(feats, dtype=torch.float32)

        label = self.class_to_idx[row["class_name"]]
        is_reject = row["class_name"] in REJECT_CLASSES
        return img, feats, label, is_reject

# --- Augmentation ---
# Rotation + flips: organisms have no canonical orientation.
# Mild brightness/contrast jitter: directly simulates the "illumination inconsistency" challenge.
# Avoid heavy color jitter: species ID leans on subtle intensity/texture cues (intensity_mass_displace etc.)
IMG_SIZE = 224

train_transform = T.Compose([
    T.Resize((IMG_SIZE, IMG_SIZE)),
    T.RandomRotation(180),
    T.RandomHorizontalFlip(),
    T.RandomVerticalFlip(),
    T.ColorJitter(brightness=0.2, contrast=0.2),
    T.ToTensor(),
    T.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
])

eval_transform = T.Compose([
    T.Resize((IMG_SIZE, IMG_SIZE)),
    T.ToTensor(),
    T.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
])
```

```python
all_rows = []
for cdir in class_dirs:
    cname = os.path.basename(cdir)
    feat_csvs = glob.glob(f"{cdir}/*.csv")
    img_files = glob.glob(f"{cdir}/*.jpeg") + glob.glob(f"{cdir}/*.jpg") + glob.glob(f"{cdir}/*.png")

    if feat_csvs:
        fdf = pd.read_csv(feat_csvs[0])
        fdf["class_name"] = cname
        # expects a filename/path column to join on — inspect if this join comes up empty
        fname_col = [c for c in fdf.columns if "file" in c.lower() or "path" in c.lower() or "image" in c.lower()]
        fdf["filepath"] = fdf[fname_col[0]].apply(lambda x: x if os.path.isabs(str(x)) else f"{cdir}/{os.path.basename(str(x))}")
        all_rows.append(fdf)
    else:
        # fallback: no feature CSV found for this class, keep images with zero-filled features
        fallback = pd.DataFrame({"filepath": img_files})
        fallback["class_name"] = cname
        all_rows.append(fallback)

full_df = pd.concat(all_rows, ignore_index=True)
full_df["basename"] = full_df["filepath"].apply(os.path.basename)
print(full_df.shape)
full_df.head()
```
